### Loading and understanding data

### Preprocessing and tokenizing the data

First we have to clean the data. For NLP, we apply various processing like converting all the words to lower case, trimming spaces, removing punctuations, etc.

```python
#this removed all punctuations, extra spaces and divides the sentence into individual words.
review_text = df['reviewText'].apply(gensim.utils.simple_preprocess)
```

### Training the Word to Vector Model

Train the model for reviews. <br>
Use a window of size 10 i.e. 10 words before the present word and 10 words ahead. <br>
A sentence with at least 2 words should only be considered, configure this using min_count parameter.

Workers define how many CPU threads to be used.<br>
The workers attribute in Gensim's Word2Vec determines how many CPU threads are used for training, allowing for parallel computation to speed up the process. It defaults to 3 but should be set to match the number of available cores (e.g., 4 or 8) for optimal performance.

For reference : <br>
https://radimrehurek.com/gensim/models/word2vec.html


## Typical Workflow 
(For Understanding)
1) **Initialize** model.Word2Vec(size=100, window=5, min_count=2, workers=4).
2) **Build Vocab:** model.build_vocab(review_text, progress_per=1000).
3) **Train:** model.train(review_text, total_examples=model.corpus_count, epochs=model.epochs)

#### Initialize the model

```python
model = gensim.models.Word2Vec(window=10, min_count=2, workers=4)
```

#### Build Vocabulary

build_vocab function builds a unique list of words (vocabulary) from the review_text corpus to prepare the model for training.
progress_per=1000 - This parameter tells the model to log progress (show updates) every 1,000 words processed

```python
model.build_vocab(review_text, progress_per=1000)
```

#### Train the Word2Vec Model

```python
model.train(review_text, total_examples=model.corpus_count, epochs=model.epochs)
```

#### Save the Model

Save the model so that it can be reused in other applications

```python
model.save("./word2vec-amazon-cell-accessories-reviews-short.model")
```
