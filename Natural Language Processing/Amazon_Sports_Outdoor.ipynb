## Exercise

Train a word2vec model on the [Sports & Outdoors Reviews Dataset](http://snap.stanford.edu/data/amazon/productGraph/categoryFiles/reviews_Sports_and_Outdoors_5.json.gz)
Once you train a model on this, find the words most similar to 'awful' and find similarities between the following word tuples: ('good', 'great'), ('slow','steady')


The dataset we are using here is a subset of Amazon reviews from the Sports & Outdoors category. The data is stored as a JSON file and can be read using pandas.

Link to the Dataset: http://snap.stanford.edu/data/amazon/productGraph/categoryFiles/reviews_Sports_and_Outdoors_5.json.gz


## Answer

```python
import pandas as pd
import gensim
```

```python
df = pd.read_json("Sports_and_Outdoors_5.json", lines=True)
df
```

```python
df.shape
```

### Preprocessing and tokenizing the data

First we have to clean the data. For NLP, we apply various processing like converting all the words to lower case, trimming spaces, removing punctuations, etc.

```python
review_text = df['reviewText'].apply(gensim.utils.simple_preprocess)
```

```python
review_text
```

```python
review_text.loc[4]
```

```python
df['reviewText'].loc[4]
```

### Training the Word to Vector Model

```python
#Initialize the model
model = gensim.models.Word2Vec(window=10, min_count=2, workers=3)
```

```python
#Build vocabulary
model.build_vocab(review_text, progress_per=500)
```

```python
#Train the model
#model.train(review_text, total_examples=model.corpus_count, epochs=model.epochs)
model.train(review_text, total_examples=model.corpus_count, epochs=model.epochs)
```

### Finding Similar Words and Similarity between words

```python
model.wv.most_similar("awful")
```

```python
model.wv.similarity(w1='good', w2='great')
```

```python
model.wv.similarity(w1='slow', w2='steady')
```
